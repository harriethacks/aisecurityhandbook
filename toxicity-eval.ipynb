# Detecting toxic language

Now let’s run our own simple evaluation that aims to classify toxicity in text using Detoxify. You can find this demo in the notebook security-eval.ipynb. Detoxify is an open-source toxicity detection model. Built on top of Hugging Face's Transformers library, it detects harmful content across categories such as general toxicity, severe toxicity, insults, and threats. We first install the detoxify package:

```python
!pip install detoxify
```

Now we import the Detoxify class from the detoxify package and loads the pre-trained original toxicity detection model into the variable model. The original model is trained on English-language data, but we can also use multilingual models.

```python
from detoxify import Detoxify

# Load the pre-trained model (original or multilingual models are available)
model = Detoxify('original')
```

We then define a list called texts, containing four example sentences that will later be evaluated for toxicity. The sentences vary in sentiment and tone to demonstrate how the Detoxify model handles diverse inputs. (Some of these I may have heard before).

```python
texts = [
    "I hate AI Security.",
    "This is a great book!",
    "Harriet is a total nerd.",
    "I hope Harriet writes a sequel"
]
```

Now we use the Detoxify model to evaluate the toxicity of each sentence in the texts list. The model’s predictions are stored in the variable results, containing scores for multiple toxicity categories, and printed.

```python
# Make predictions
results = model.predict(texts)

# Display neatly
import pandas as pd
df_results = pd.DataFrame(results, index=texts)
print(df_results)
```
